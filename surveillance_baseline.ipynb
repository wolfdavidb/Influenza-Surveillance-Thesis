Now that I have combined geolocation, populations, counties, and facilities all by FIPS I will compute great-circle distances from each sequencing site. I will use county centroid coordinates for each of the sites as their location. This does cause some error as the sequencing centers are not necessarily in the center of their county, but this is acceptable when looking at national level sequencing coverage.

```python
# 1) convert degree coordinates to radians
county_coords_rad = np.radians(counties_df[["latitude", "longitude"]].to_numpy())
facility_coords_rad = np.radians(facilities_merged[["latitude", "longitude"]].to_numpy())

# 2) vectorized haversine formula
def compute_haversine_matrix(coords_a, coords_b):
    """
    coords_a: (n, 2) array of [lat, lon] in radians (counties)
    coords_b: (m, 2) array of [lat, lon] in radians (facilities)
    returns: (n, m) pairwise distance matrix in kilometers
    """
    r = 6371.0  # earth mean radius in kilometers
    
    lat1, lon1 = coords_a[:, 0:1], coords_a[:, 1:2]
    lat2, lon2 = coords_b[:, 0:1].T, coords_b[:, 1:2].T
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = np.sin(dlat / 2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2.0)**2
    c = 2.0 * np.arcsin(np.clip(np.sqrt(a), 0.0, 1.0))
    
    return r * c

# 3) generate distances
baseline_dist_matrix = compute_haversine_matrix(county_coords_rad, facility_coords_rad)

# 4) find index and distance to nearest facility for each county
nearest_facility_idx = np.argmin(baseline_dist_matrix, axis=1)

counties_df["baseline_min_dist_km"] = np.min(baseline_dist_matrix, axis=1)
counties_df["nearest_facility_id"] = facilities_merged.iloc[nearest_facility_idx]["facility_id"].values
counties_df["nearest_facility_name"] = facilities_merged.iloc[nearest_facility_idx]["name"].values
counties_df["nearest_facility_tier"] = facilities_merged.iloc[nearest_facility_idx]["tier"].values

print(f"matrix shape: {baseline_dist_matrix.shape}")
print(f"mean distance across counties: {counties_df['baseline_min_dist_km'].mean():.2f} km")
print(f"min distance: {counties_df['baseline_min_dist_km'].min():.2f} km")
print(f"max distance: {counties_df['baseline_min_dist_km'].max():.2f} km")
```

Now that I have distances from each county to the nearest sequencing site, I will population weighted distances.

```python
# 1) population-weighted mean distance
total_us_pop = counties_df["pop_2024"].sum()
pop_weighted_mean_km = np.average(counties_df["baseline_min_dist_km"], weights=counties_df["pop_2024"])

print(" baseline spatial coverage metrics (n=57 facilities) ")
print(f"total us 2024 population modeled: {total_us_pop:,}")
print(f"county-unweighted mean distance:   {counties_df['baseline_min_dist_km'].mean():.2f} km")
print(f"population-weighted mean distance: {pop_weighted_mean_km:.2f} km")
print(f"county-unweighted median distance: {counties_df['baseline_min_dist_km'].median():.2f} km")

# 2) catchment coverage thresholds
thresholds_km = [25, 50, 100, 150, 200]
coverage_records = []

for radius in thresholds_km:
    mask = counties_df["baseline_min_dist_km"] <= radius
    cov_counties = mask.sum()
    cov_pop = counties_df.loc[mask, "pop_2024"].sum()
    
    pct_counties = (cov_counties / len(counties_df)) * 100.0
    pct_pop = (cov_pop / total_us_pop) * 100.0
    
    coverage_records.append({
        "catchment_radius_km": radius,
        "counties_covered": cov_counties,
        "pct_counties": round(pct_counties, 2),
        "population_covered": cov_pop,
        "pct_population": round(pct_pop, 2)
    })

baseline_coverage_df = pd.DataFrame(coverage_records)
display(baseline_coverage_df)

# 3) 5 most isolated counties
print("\n top 5 most isolated counties (maximum distance to sequencing) ")
display(
    counties_df.sort_values(by="baseline_min_dist_km", ascending=False)[
        ["fips", "county_name", "state_name", "pop_2024", "baseline_min_dist_km", "nearest_facility_id"]
    ].head(5)
)
```
